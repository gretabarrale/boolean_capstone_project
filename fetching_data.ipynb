```python
import pandas as pd
import os
import time
import requests
```

```python
istat_csv = 'istat_dataset.csv'
url = 'https://esploradati.istat.it/SDMXWS/rest/data/41_983'
headers =  {'Accept': 'application/vnd.sdmx.data+csv;version=1.0.0'}
if os.path.exists(istat_csv):
    istat_df = pd.read_csv(istat_csv, dtype={"REF_AREA": str})
else:
    counter= 0 
    while counter < 5:
        try:
            response = requests.get(url, headers=headers)
            response.raise_for_status()
            with open(istat_csv, 'w', encoding='utf-8') as file:
                file.write(response.text)
            istat_df = pd.read_csv(istat_csv, dtype={"REF_AREA": str})
            print(f"Data saved to {istat_csv}")
            break
        except requests.exceptions.RequestException as error:
            print(f"Error: {error}")
            counter = counter + 1
            time.sleep(20)
            if counter == 5:
                raise Exception("Fetching data failed")
```

##  OBS_VALUE, RESULT AND DATA_TYPE


**OBS_VALUE** was confirmed to be a numeric count (of injured, killed people or road accidents) based on:
-  dtype being `int64` (confirmed via `.info()`), meaning whole numbers with no decimals 
- `.describe()` showing whole numbers with and a reasonable range and no decimals (except for mean and std which is expected)
- `UNIT_MEAS` and `UNIT_MULT` both being NaN, meaning no unit conversion or
  scaling is needed before summing the values

This confirms that `OBS_VALUE` can be directly summed or compared across rows

**RESULT** was confirmed thanks to the official ISTAT codelist (CL_ESITO), available at:
`https://esploradati.istat.it/SDMXWS/rest/codelist/IT1/CL_ESITO`

- **M** = "killed" — not male gender (as initially assumed)
- **F** = "injured" — not female gender (as initially assumed)
- **9** = "total"

This explains why:
- **KILLINJ** (killed and injured) only shows M and F, never 9 so the total number of injured and killed people must be calculated by summing F+M manually
- **ROADACC** (road accidents) only shows 9, since a road accident is a single event and doesn't split into killed/injured like KILLINJ does


```python
#Dropping empty columns from istat_df (mostly metadata)
dropped_columns_istat_df = ['OBS_STATUS', 'NOTE_DS', 'NOTE_REF_AREA', 'NOTE_DATA_TYPE', 'NOTE_RESULT', 'NOTE_TIME_PERIOD', 'BASE_PER', 'UNIT_MEAS', 'UNIT_MULT']
istat_df = istat_df.drop(columns= dropped_columns_istat_df)
istat_df.info()
```

```python
# Drop DATAFLOW and FREQ columns: both of them are constant across all rows and add no informations to the analysis
istat_df = istat_df.drop(columns=['DATAFLOW', 'FREQ'])
istat_df.info()
```

```python
#Checking the unique values of situas_df's columns to confirm whether they add relevant info to the analysis
for col in ['Codice Ripartizione geografica', 'Codice Regione', 'Codice Provincia (Storico)', 'Comune (dizione straniera)', 'Sigla automobilistica', 'Capoluogo di Provincia/Uts', 'Capoluogo di Regione', 'Popolazione legale', 'Anno Censimento']:
    print(col)
    print(situas_df[col].unique())
    print()
```

```python
#Dropping unnecessary situas's columns for the analysis
dropped_columns_situas_df = ['Codice Ripartizione geografica', 'Codice Regione', 'Codice Provincia (Storico)','Codice Provincia/Uts', 'Codice Comune (numerico)', 'Comune (dizione straniera)', 'Capoluogo di Provincia/Uts', 'Capoluogo di Regione', 'Popolazione legale', 'Anno Censimento']
situas_df = situas_df.drop(columns= dropped_columns_situas_df)
situas_df.info()
```

```python
#Pivot istat_df to one row per municipality/year, obtaining separate columns for accidents/killed/injured.
# Using sum as aggregation because OBS_VALUE is a count of people/events. Summing values preserves the actual total since each row contains one value.

istat_df_pivot = istat_df.pivot_table(
    index=['REF_AREA', 'TIME_PERIOD'],
    columns=['DATA_TYPE', 'RESULT'],
    values='OBS_VALUE',
    aggfunc='sum'
).reset_index()
istat_df_pivot.head(10)
```

```python
#Checking columns order
istat_df_pivot.columns.tolist()
```

```python
# Flatten and rename the two-level pivot columns 
istat_df_pivot.columns = ['REF_AREA', 'TIME_PERIOD', 'INJURED', 'KILLED', 'TOTAL_ACCIDENTS']
istat_df_pivot.head()
```

```python
# Joined istat_df_pivot with cleaned situas_df. Using two keys per df in order to avoid duplicates and misleading data.
# Using inner join since situas_df only covers 2018-2024: this naturally filters istat_df_pivot, leading to the year range selected for the analysis.
df = istat_df_pivot.merge(situas_df,
    left_on=['REF_AREA', 'TIME_PERIOD'],
    right_on=['Codice Comune (alfanumerico)', 'Anno (Popolazione residente)'],
    how='inner')
df.sample(5)
```
